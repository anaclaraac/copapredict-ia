```python
import pandas as pd

matches = pd.read_csv(
    "../data/raw/matches.csv"
)

matches.head()
```

```python
matches.duplicated().sum()
```

```python
matches.isnull().sum()
```

```python
matches["home_team"] = (
matches["home_team"]
.str.strip()
.str.title()
)

matches["away_team"] = (
matches["away_team"]
.str.strip()
.str.title()
)
```

```python
matches["date"] = pd.to_datetime(
matches["date"]
)
```

```python
matches["year"] = (
matches["date"]
.dt.year
)
```

```python
matches["month"] = (
matches["date"]
.dt.month
)
```

```python
matches.to_csv(
"../data/processed/matches_clean.csv",
index=False
)
```
