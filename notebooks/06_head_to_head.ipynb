```python
import pandas as pd
import numpy as np

matches = pd.read_csv(
"../data/processed/context_model_input.csv"
)
```

```python
matches = matches.sort_values(
"date"
)
```

```python
matches[
"h2h_home_wins"
] = 0

matches[
"h2h_away_wins"
] = 0

matches[
"h2h_draws"
] = 0
```

```python
for i in range(
len(matches)
):

    casa = matches.loc[
    i,
    "home_team"
    ]

    fora = matches.loc[
    i,
    "away_team"
    ]

    historico = matches.iloc[:i]

    jogos = historico[
        (
        (
        historico[
        "home_team"
        ] == casa
        )

        &

        (
        historico[
        "away_team"
        ] == fora
        )
        )
    ]

    casa_v = (
    jogos[
    "home_score"
    ]
    >
    jogos[
    "away_score"
    ]
    ).sum()

    fora_v = (
    jogos[
    "away_score"
    ]
    >
    jogos[
    "home_score"
    ]
    ).sum()

    empates = (
    jogos[
    "home_score"
    ]
    ==
    jogos[
    "away_score"
    ]
    ).sum()

    matches.loc[
    i,
    "h2h_home_wins"
    ] = casa_v

    matches.loc[
    i,
    "h2h_away_wins"
    ] = fora_v

    matches.loc[
    i,
    "h2h_draws"
    ] = empates
```

```python
matches.to_csv(
"../data/processed/h2h_model_input.csv",
index=False
)
```
