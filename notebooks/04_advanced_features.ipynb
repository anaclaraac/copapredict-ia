```python
import pandas as pd
import numpy as np
```

```python
matches = pd.read_csv(
"../data/processed/model_input.csv"
)
```

```python
matches.head()
```

```python
matches[
"home_defense_avg"
] = (

matches[
"away_score"
]

.rolling(5)

.mean()

)
```

```python
matches[
"away_defense_avg"
] = (

matches[
"home_score"
]

.rolling(5)

.mean()

)
```

```python
matches[
[
"home_defense_avg",

"away_defense_avg"
]
].head()
```

```python
matches["away_form"] = (
    matches["away_score"] >
    matches["home_score"]
).astype(int).rolling(5).mean()
```

```python
matches[
"home_win"
] = (

matches[
"home_score"
]

>

matches[
"away_score"

]

).astype(
int
)
```

```python
matches[
"win_streak"
] = (

matches[
"home_win"
]

.rolling(
5
)

.sum()

)
```

```python
matches[
"home_loss"
] = (

matches[
"home_score"
]

<

matches[
"away_score"

]

).astype(
int
)
```

```python
matches[
"loss_streak"
] = (

matches[
"home_loss"
]

.rolling(
5
)

.sum()

)
```

```python
matches.columns
```

```python
matches["home_attack_avg"] = (
    matches[
        "home_score"
    ]
    .rolling(5)
    .mean()
)
```

```python
matches[
[
"home_score",
"home_attack_avg"
]
].head(10)
```

```python
matches["away_attack_avg"] = (
    matches[
        "away_score"
    ]
    .rolling(5)
    .mean()
)
```

```python
matches[
[
"away_score",
"away_attack_avg"
]
].head(10)
```

```python
matches["home_form"] = (
    (
        matches[
            "home_score"
        ]
        >
        matches[
            "away_score"
        ]
    )
    .astype(int)
    .rolling(5)
    .mean()
)
```

```python
matches[
[
"home_form"
]
].head(10)
```

```python
print(
matches.columns
)
```

```python
matches = matches.replace(
[np.inf,-np.inf],
np.nan
)
```

```python
matches = matches.dropna(
subset=[

"home_attack_avg",

"away_attack_avg",

"home_defense_avg",

"away_defense_avg",

"home_form",

"away_form",

"win_streak",

"loss_streak",

"target"

]
)
```

```python
matches.isnull().sum()
```

```python
matches.to_csv(

"../data/processed/advanced_model_input.csv",

index=False

)
```
